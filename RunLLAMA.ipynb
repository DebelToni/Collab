```python

from langchain_core.prompts import ChatPromptTemplate
from langchain_ollama.llms import OllamaLLM

template = """Question: {question}

Answer: Let's think step by step."""

prompt = ChatPromptTemplate.from_template(template)

model = OllamaLLM(model="llama3.2:3b")

chain = prompt | model

chain.invoke({"question": "What is MoE in AI?"})
```

```python
import os
from openai import OpenAI

def main():
    # 1. Configure your API key (you can also set OPENAI_API_KEY in your env)

    # 2. Instantiate the OpenAI client
    client = OpenAI()

    # 3. Call the non-chat completions endpoint
    resp = client.completions.create(
        model="llama3.2:3b",
        prompt="Please summarize the key benefits of using Python for data science.",
        max_tokens=150,
        temperature=0.7
    )

    # 4. Print out the raw continuation
    text = resp.choices[0].text
    print("Model output:", text.strip())

if __name__ == "__main__":
    main()
```
